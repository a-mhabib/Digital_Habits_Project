 3-Check Missing Values

```python
# Missing Value

df.isna().sum()
```

```python
# Handle Missing value in perceived_productivity_score col throug drop rows  because the prenctage  = 5% 

df.dropna(subset='perceived_productivity_score',inplace=True )
```

```python
df.reset_index(drop=True)
```

```python
# Fill missing value in daily_social_media_time col using median (Outliers)

median_daily_time=round(df['daily_social_media_time'].median(),2)
df['daily_social_media_time']=df["daily_social_media_time"].fillna(median_daily_time)

```

```python
# Fill misiing values in Stress_level col using mean because it float and no outliers
mean_stress_level=round(df['stress_level'].mean(),1)
df["stress_level"]=df["stress_level"].fillna(mean_stress_level)
```

```python
# Fill missing valuse in sleep_hours col using mean (Float and no outliers)
mean_sleep_hours=(round(df['sleep_hours'].mean(),2))
df['sleep_hours']=df['sleep_hours'].fillna(mean_sleep_hours)
```

```python
df.isna().mean( ).round(2) * 100
```

```python
# I am modfied this becaue the description of this col say in hours 
df["screen_time_before_sleep"]=(df["screen_time_before_sleep"].round(1))
```

```python
df["screen_time_before_sleep"].head()
```

```python
# Fill missing valuse in screen_time_before_sleep col using mean (Float and no outliers)
mean_screen_time_before_sleep=(round(df['screen_time_before_sleep'].mean(),1))
df['screen_time_before_sleep']=df['screen_time_before_sleep'].fillna(mean_screen_time_before_sleep)
```

```python
# Fill missing valuse in actual_productivity_score col using mean (Float and no outliers)
mean_actual_productivity_score=(round(df['actual_productivity_score'].mean(),2))
df['actual_productivity_score']=df['actual_productivity_score'].fillna(mean_actual_productivity_score)
```

```python
df.isna().mean( ).round(2) * 100
```

#### Digital habits => Productivity

Q1 Do focus apps affect productivity?
